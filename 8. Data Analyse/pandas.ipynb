```python
## Series

# A Pandas series is one-dimentional array-like object that can hold any data type. It is similar to a column in  a table 

import pandas as pd
data=[1,2,3,4,5]
series=pd.Series(data)
print(series)
print(type(series))  # here In Series the 0 to 4 is index and the 1 to 5 is element. Basically the 0 index assign=1, 1 index assign=2, Similerley all 3 to 5 have indexing.
```

```python
# Create a series from dictionery

data={'a':1,'b':2,'c':3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data=[10,20,30]
index=['a','b','c']

series=pd.Series(data,index)
print("Series \n",series)
print(type(series))
```

```python
data={
    'Name':["Krish","Rohit","Amar"],
    'age':[32,25,30],
    'City':["Vadala","Kamothe","Kalyan"]

}

df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## Create a Dataframe from a List of Dictionery

data=[
    {'Name':"Krish",'Age':32,'City':"Vadala"},
    {'Name':"John",'Age':34,'City':"Vadala"},
    {'Name':"Rohit",'Age':30,'City':"Vadala"},
    {'Name':"Amar",'Age':32,'City':"Vadala"}
]

df=pd.DataFrame(data)
print(df)
print(type(df))
```
