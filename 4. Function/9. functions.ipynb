```python
## Variable lenght argument

## Positional and keyword arguments

## 1- Positional Argument

def print_number(*args):  ## the * refering the all the postions argument of paramater. Means mutiple parameter. if you want to use multiple parameter of printing the use *(parameter).
        for number in args:
                print(number)
```

```python
print_number(1,2,3,4,5,6,"Rohit")
```

```python
## 2- Keyword Arguments ## In keyword argument all the parameter in the form of key_value arguments

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="rohit",age="32",city="panvel")
```

```python
def print_details(*args,**kwargs):
    for val in args:
                print(f" Positional Argument : {val}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,"rohit",name="Rohit",age="32",city="panvel")
```
