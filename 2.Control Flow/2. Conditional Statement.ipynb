```python
## Nested Conditional Statement

# You can place if,else,elif statement inside another if, elif,else statement to create nested Conditional Statement.

num=int(input("Enter the Number:"))

if num>0:
    print("The number is positive")
    if num%2==0:
        print("This number is even")
    else:
        print("The number is odd.")
elif num==0:
    print("The number is zero.")
else:
    print("The Number is Negative")
    if num%-2==0:
        print("The number is Even Negative")
    else:
        print("The number is odd negative.")
```

```python
## practice Example

#   Determine if a year is a leap year using nested condition statement.

Year=int(input("Enter the Year:"))

if Year%4==0:
    if Year%100==0:
        if Year%400==0:
            print(f" This is Leaf year:{Year}")
        else:
            print("This is not leap year:",Year)
    else:
        print("is a leap year",Year)

else:
    print("is not a leap year",Year)
   
```

```python
## Assignement 
# Simple Calculator Program. by If, elif,else.

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
operation = input("Enter operation (+, -, *, /): ")


#result_add= num1 + num2
#result_sub= num1 - num2
#result_multi= num1 - num2
#result_div= num1 / num2
if operation=='+':
    result = num1 + num2
elif operation== '-':
    result = num1 - num2
elif operation== '*':
    result = num1 * num2
elif operation== '/':
    if operation!=0:
        result= num1 / num2

else:
    print("The result is Invalid")

print(f"Result = {num1} {operation} {num2}:",result)

```
