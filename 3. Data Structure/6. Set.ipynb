```python
## Create a Set

my_set={1,2,3,4,5}

print(type(my_set))
```

```python
my_set=set([1,2,3,4,5])
print((my_set))
```

```python
my_empty_set=set([1,2,3,6,5,4,5])
print((my_empty_set))
```

```python
## Basic Set Operation

## Adding and removing element

my_set.add(7)
print(my_set)
my_set.add(7)
print(my_set)
```

```python
## Remove the Element set

my_set.remove(3)
print(my_set)
```

```python
my_set.remove(10)
print(my_set)           ## The 10 Element not store in the variable
```

```python
my_set.discard(10)      ## discard means remove an element from set if it is present if not it not gives error.
print(my_set)
```

```python
## Pop method
remove_element=my_set.pop()     ## In pop method what ever element in first is first in first out. That's why 1 is remove and it give {2,4,5,7}
print(remove_element)
print(my_set)
```

```python
## Clear all the element

my_set.clear()
print(my_set)
```

```python
## Set membership test

my_set={1,2,3,4,5}              # It's gives True or False if it's Present or not
print(3 in my_set)
print( 10 in my_set)
```

```python
### Mathematical Operation

set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

## Union
union_set=set1.union(set2)          ## union is adding 2 set of variable
print(union_set)


## Intersection

intersection_set1=set1.intersection(set2)           # Intersection give the repeated value of sets.
print(intersection_set1)                        

set1.intersection_update(set2)      ## Repated/ common element are updated in set1 all the element which not repated they are eraise and printed the common/repated element
print(set1)


```

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

## Difference

print(set1.difference(set2))
```

```python
set2.difference(set1)
```

```python
## Symmentric Difference 

set1.symmetric_difference(set2)     # WHich are the Common element in set1 and set2 this are be removed only.
```

```python
### Set methods

set1={1,2,3,4,5}
set2={3,4,5}

# is_subset means what are element in set1 are present it in set2 or not then it gives in True or false

print(set1.issubset(set2))

print(set1.issuperset(set2))
```

```python
## Counting unique words in text

text="In this tutorial we are discussing about sets"
words=text.split()

## Convert list of words to set to get unique words

unique_words=set(words)
print(unique_words)
print(len(unique_words))
```
