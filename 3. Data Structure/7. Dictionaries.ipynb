```python
## In Dictioneries always put value in key value pairs

student={"name":"rohit","age":23,"Grade":23}
print(student)
print(type(student))
```

```python
## single key is always use

student={"name":"rohit","age":23,"name":23}     # When you defined two keys as same and value is put it difference then print the key and the last value you defined is display.
print(student)
```

```python
## Accessing dictionery element

student={"name":"rohit","age":23,"Grade":"A"}
print(student)
```

```python
## Accessing Dictionery element
print(student['Grade'])
print(student['age'])
print(student['name'])

## Accessing using get() method

print(student.get('Grade'))         ## Accesing the value by get method of which already defined key
print(student.get('last_name'))     ## Not Accessing the value by get method bcoz key not defined .
print(student.get('last_name','not applicable'))    ## if we haven't key then also we give some condition if not then print this.
```

```python
## Modifying Dictionery element

## Dictionery are mutable, so you can add, update or delete elements

print(student)
```

```python
student["age"]=33           ## Updated value for the key
print(student)
student["Address"]="india"      ## Adding new key and value
print(student)
```

```python
del student["Grade"]        ## delete key and value pair

print(student)
```

```python
## Dictionery methods

keys=student.keys()     ## getting all the keys
print(keys)
values=student.values()      ## getting all the values
print(values)


items=student.items()       ## All key value pairs
print(items)
```

```python
## Shallow copy

student_copy=student
print(student)
print(student_copy)
```

```python
student["name"]="rohit1"
print(student)
print(student_copy)         ## In This when i update student it also automatically update the student_copy. there is problem that's why we use shallow copy
```

```python
student_copy=student.copy()     ## Shallow copy, means when we changed the variable key- value it's not effect on other variable
print(student_copy)
print(student)
```

```python
student["name"]="rohit2"
print(student_copy)
print(student)
```

```python
### Iterating over dictioneries
## you can use loop to iterate over dictioneries, key ,values,or items


## Iterating over keys
for keys in student.keys():
    print(keys)

```

```python
## Iterate over values
for values in student.values():
    print(values)
```

```python
## Iterate over key value pairs
for key,value in student.items():
    print(f'{key}:{value}')
```

```python
## Nested Dictioneries

students={
    "student1":{"name":"Rohit","age":32},
    "student2":{"name":"rahil","age":35}

}
print(students)
```

```python
## Access nested dictionery element

print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
### Iterating over nested dictioneries

for student_id,student_info in students.items():
    print(f'{student_id}:{student_info}')
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionery comprehension

squres={ x:x**2 for x in range(5)}
print(squres)
```

```python
## Condition dictionery comprehension

even1={x:x**2 for x in range(5) if x%2==0}
print(even1)                                        ## Squers of even numbers
```

```python
## Practical Example

## use a dictionery to count the frequency of element in list

numbers=[1,2,2,3,3,3,4,4,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
### Marge 2 dictionaries into one

dict={"a":1,"b":2}
dict1={"b":3,"c":4}
merged_dict={**dict,**dict1}
print(merged_dict)
```
