```python
## List comprehension

lst=[]
for x in range(10):
    lst.append(x**2)  # In the range 1 to 9 it gives squers of each Number. Like that- 2*2=4, 3*3=9, 9*9=81

print(lst)
```

### List comprehension

Basic Syntax:               [expression for item in iterable]

with conditional logic      [expression for item in iterable if condition]

Nested List Comprehension   [expression for item1 in iterable for item2 in iterable2]

```python
## Basic List comphrension

sqaure=[num**2 for num in range(10)]
print(sqaure)
```

```python
## List Compresion with condition
list=[]
for i in range (10):
    if i%2==0:
        list.append(i)

print(list)
```

```python
even_number=[num for num in range(10) if num%2==0]
print(even_number)
```

```python
### Nested List Comprehension

lst1=[1,2,3,4]
lst2=['a','b','c','d']

pair=[[i,j]for i in lst1 for j in lst2]

print(pair)
```

```python
## List comprehension with function calls

Words=["hello","world","python","list","comprehension"]
lengths= [len(Word) for Word in Words]

print(lengths)
```
