### Logical Operators

```python
#  And, Or, Not

# And Logical Operator - If Both variable are True then It's True either It's false
# Or Logical Operator - If Both variable are False then It's False either It's True
# Not Logical Operator - It Give reverse Answer. if It's true but in Not it's false.   if It's false but in Not it's True.

# And Example

x= True
y= True

result= x and y
print(result)     # It's true

# But if One variable are false then it's false.

x= False
y= True

result= x and y
print(result)    # Then It's false
```

```python
# Or Example

x= False
y= False

result= x or y
print(result)     # It's get false

# But if one variable are True then it's True

x= False
y= True

result= x or y
print(result)
```
